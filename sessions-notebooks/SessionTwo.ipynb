## 1. Theory: Conditional Expressions (Ternary Operator)

In Python, a conditional expression (often called a ternary operator) allows you to assign a value to a variable or return a value based on a condition in a single, compact line.

**Syntax:**
```python
variable = value_if_true if condition else value_if_false
```
You can also chain multiple conditions using `elif`-like structures: `val1 if cond1 else val2 if cond2 else val3`.

### Example: If-Else Statement inside normal variable assignment

````python
# Define a name variable
name = 'Faisal'

# Inline conditional assignment evaluating if name matches 'Faisal'
check = 'You are a student' if name == 'Faisal' else 'You are not a student'
````

````python
print(check)
````

````python
def count(number):
    """
    Check whether a given number is bigger than, equal to, or lesser than 5.
    
    Parameters:
    number (int/float): The numeric value to evaluate.
    
    Returns:
    str: A descriptive string of the comparison result.
    """
    # Chaining conditional expressions for multiple branches
    check = 'Bigger than 5' if number > 5 else 'Equal to 5' if number == 5 else 'Lesser than 5'
    return check
````

````python
# Test the count function with 5
count(5)
````

````python
'''
Grading scale logic:
- degree > 90 => A
- degree > 80 => B
- degree > 60 => C
- otherwise   => F
'''
# Prompt user input and convert to integer
dgree = int(input('Enter your dgree: '))

# Assign grade using nested inline conditional statements
check = 'A' if dgree > 90 else 'B' if dgree > 80 else 'C' if dgree > 60 else 'F'
````

## 2. Theory: List Comprehension

List comprehension is a concise and elegant way to create lists in Python. It consists of brackets containing an expression followed by a `for` clause, then zero or more `for` or `if` clauses.

**Syntax:**
```python
[expression for item in iterable if condition]
```

````python
# using for loop inside lists is just like using for loop to append elements to list

# let's assign a p to be an empty list
p = []
# now I'll use for loop to append elements to p
for i in range(1, 11):
    p.append(i)
print(p)
````

````python
# using list comperhension is so useful and we'll see that father, so let's get used to use it
l = [i for i in range(1,11)] # this line does what the cell above does
print(l)
````

````python
# also I can do type Casting to the element as I can when doing a for loop
a = []
for i in range(1,11):
    a.append(str(i))
print(a) 
````

````python
# Define a tuple containing pairs of numbers
t = ((1, 3), (4, 6), (8, 10))
````

````python
def double(num): 
    """
    Multiplies a given number by 2.
    
    Parameters:
    num (int/float): The input number.
    
    Returns:
    int/float: The doubled value.
    """
    return num * 2
````

### Example: List comprehension

````python
# Apply list comprehension: unpack tuple elements (i, a), double i, and cast a to float
'''
see that we're calling double on (i)
'''
x = [(double(i), float(a)) for i, a in t]
print(x)
````

## 3. Theory: Lambda Functions

A lambda function is a small anonymous function defined with the `lambda` keyword. It can take any number of arguments, but can only have one expression. The expression is evaluated and returned.

**Syntax:**
```python
lambda arguments: expression
```

### Example: Lambda function

````python
# Standard function definition for comparison
def double(x):
    """Return double of x using a standard def function."""
    return x * 2
````

````python
# Call the standard function
a = double(4)
print(a)
````

````python
# Equivalent lambda function assigned to a variable
double = lambda x: x * 2
````

````python
# Call the lambda function
double(5)
````

````python
# Raw scores stored in a list
raw_scores = [50, 65, 80, 95] 
````

````python
def add_score(score):
    """
    Adjusts a student's score. Scores >= 80 remain unchanged,
    while lower scores get a +5 bonus.
    
    Parameters:
    score (int): The original score.
    
    Returns:
    int: The adjusted score.
    """
    if score >= 80:
        return score
    score += 5
    return score
````

## 4. Theory: Function Docstrings & Comments

* **Comments (`#`)** are used to explain specific lines or blocks of code for developers reading the source code. Python ignores them at runtime.
* **Docstrings (`"""..."""`)** are multi-line string literals placed right after a function, class, or module definition. Unlike comments, docstrings are retained at runtime and can be accessed via `function_name.__doc__` or `help(function_name)`.
