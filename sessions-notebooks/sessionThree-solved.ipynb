```python
# creat the table class
class table:
    
    '''
    This class returns a formatted table
    from given headers and row_data
    both headers and row_data should be lists
    the class has 3 methods:
        set_headers -> this method set the headers values
        add_row_data -> this method add values to the row_data attribute
        make_talbe -> this method returns the table
    '''
    
    
    
    """define the __init__ method"""
    def __init__(self):
        self.headers = []
        self.row_data = []
    
    
        
    """define the method that sets the values of the headers attribute"""
    def set_headers(self, headers: list):
        # check if the input is list
        if isinstance(headers, list) == False:
            print('Error: The Input Is Not List')
            return
        # add the input to the headers attribute (type cast it to be str)
        self.headers = [str(h) for h in headers]
        
        
    """define the method that adds values to the row_data attribute"""
    def add_row_data(self, data: list):    
        # check if the input is list
        if isinstance(data, list) == False:
            print('Error: The Input Is Not List')
            return
        # check if the input's length matching the headers attribute length
        if len(data) != len(self.headers):
            print('Error: The Length Is Not Accepted')
            return
        # add the input to the row_data attribute (type cast it to be str)        
        self.row_data.append([str(d) for d in data])

    
    
    """define the method that makes the table"""
    def make_table(self):    
        # 1- the width:
        # set the width to be taking the len from the header
        width = [len(h) for h in self.headers]        
        # do a nested for loop to compare the length in header and row_data (use: enumerate, max) 
        for row in self.row_data:
            for i, val in enumerate(row):
                width[i] = max(width[i], len(val))
        
        # 2- formate the header section:
        # do a for loop to do the alignment and the format (use: enumerate, ljust)
        h_cell = []
        for i, h in enumerate(self.headers):
            h_cell.append(f' {h.ljust(width[i])} ')
        
        # then do: '|' + "|".join() + '|' for the headers
        h_line = '|' + '|'.join(h_cell) + '|'
       
        
        # 3- create the boareder:
        boarder = '+' + '+'.join(['-' * (w + 2) for w in width]) + '+'
        # 4- create a list that holds the formatted desgin for you:
        design = [boarder, h_line, boarder]
        
        # 5- formate the row section:
        # do a nested for loop to do the alignment and the format (use: enumerate, ljust)
        for row in self.row_data:
            r_cell = []
            for i, r in enumerate(row):
                r_cell.append(f' {r.ljust(width[i])} ')
            # then do: '|' + "|".join() + '|' for the rows
            r_line = '|' + '|'.join(r_cell) + '|'
            # append the formatted section to the list that holds the design for you then append the boarder
            design.append(r_line)
            design.append(boarder)
        
        # 6- retrun the table:
        # retrun the list that holds the disgin for you (use: '/n'.join())
        return '\n'.join(design)
```

```python
table = table()
table.set_headers(['Name', 'Age', 'Salary', 'Role'])
table.add_row_data(['Faisal Abdallatif', 20, 1000, 'Admin'])
print(table.make_table())
```
